#ARCHITECTURE:-

CHATBOT --> INTENT --> ENTITY -->FULFILMENT

1. using dialog flow--> chatbot building framework
2. user interacts with this chatbot
3. user does a query, example:- what is the wheather today? --> that is intent in chatbot's language
4. intent is seomthing that user wants chatbot to do
5. We extract entity from the intent (chatbot does that). entities can be more than one
6. for example:- user asks can you provide me your phone number i need help. here this whole thing is an intent, and phone number here is an entity.
7. Another example:- user asks will it rain today, this whole is an intent and "rain" is an entity.
8. Then fulfilment --> here we make a flask app/ api
9. this flask api does the thing, example:- if user is asking phone number so it will search the phone number and sent to chatbot
10. then chatbot as reply presents to the user.
11. this chatbot can be anywhere in whatsapp, telegram etc etc
